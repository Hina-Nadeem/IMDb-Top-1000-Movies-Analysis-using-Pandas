```python
imdb.drop_duplicates()
```

```python
if 'IMDB_Rating' in imdb.columns:
    imdb['IMDB_Rating']= imdb['IMDB_Rating'].fillna(imdb['IMDB_Rating'].mean())
```

```python
imdb['IMDB_Rating']
```

```python
if 'Runtime' in imdb.columns:
    imdb['Runtime'] = imdb['Runtime'].astype(str)
    imdb['Runtime'] = imdb['Runtime'].str.extract(r'(\d+)').astype(int)
```

```python
imdb.Runtime
```

```python
if 'IMDB_Rating' in imdb.columns:
    print(imdb['IMDB_Rating']>8.5)
```

```python
high_rating=imdb[imdb['IMDB_Rating']>8.5] #a filter or a small dataset 
print(high_rating['IMDB_Rating'].head())
```

```python


# Now filter rows that contain both "Action" and "Sci"
movie = imdb[
    imdb['Genre'].str.contains('Action', case=False) & #So the filter becomes case-insensitive bcz of case=false
    imdb['Genre'].str.contains('Sci-Fi', case=False)
]

print(movie[['Series_Title', 'Genre']].head())#So, yes — movie mein sare columns hain jo imdb mein the, 
#bas aap ne unko filter kiya hai specific genres ke liye.
```

```python
# Convert 'Released_Year' to numeric (integer), if not already
imdb['Released_Year'] = pd.to_numeric(imdb['Released_Year'], errors='coerce')#This line ensures that all values in the
#'Released_Year' column are numeric so you can use comparison operations (>=, <=, etc.) without running into errors. And,
#if there are any invalid values, they won't break your code, they’ll just be marked as NaN.the purpose of errors

# Filter the data for movies released between 2000 and 2015
movies = imdb[(imdb['Released_Year'] >= 2000) & (imdb['Released_Year'] <= 2015)]

# Print the selected columns
print(movies[['Series_Title', 'Released_Year']])
```

```python
sorted_imdb = imdb.sort_values(by='IMDB_Rating', ascending=False)
print(sorted_imdb[['Series_Title', 'IMDB_Rating']].head())
```

```python
avg_rating = imdb.groupby('Genre')['IMDB_Rating'].mean()
print(avg_rating)
```

```python
most_movies_year = imdb['Released_Year'].value_counts().idxmax()
#value_counts()	Har unique value ka count batata hai. ek year kitni bar araha 
#idxmax()	Sabse zyada count wali value return karta hai

print(most_movies_year)
```
